## Initial Findings

- The dataset contains 10,000 observations and 14 variables.
- No missing values were detected.
- No duplicated rows were detected.
- Machine failures represent only 3.39% of the dataset, indicating a significant class imbalance.
- Product types are distributed as L (60.00%), M (29.97%), and H (10.03%).
- UID and Product ID are identifiers and will not be used as predictive features.
- TWF, HDF, PWF, OSF and RNF directly describe failure modes and will be excluded from model inputs to prevent target leakage.
- A small inconsistency exists between the overall machine-failure label and individual failure-mode indicators.
- Product Type is categorical and will require encoding.
- Evaluation metrics such as Recall, Precision, F1-score and PR-AUC will be more informative than accuracy alone.

### Target Consistency Check

A consistency check was performed between the `Machine failure` target and
the individual failure indicators (`TWF`, `HDF`, `PWF`, `OSF`, `RNF`).

Results showed that:

- 330 machine failure observations contain at least one explicit failure flag.
- 18 observations contain a failure flag while `Machine failure` is 0.
- 9 observations are labeled as machine failures although none of the five
  failure indicators are active.

These observations indicate a small amount of label inconsistency or noise
in the dataset. The original `Machine failure` column will be retained as
the prediction target.
